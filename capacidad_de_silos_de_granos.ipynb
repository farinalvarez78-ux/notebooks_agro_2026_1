#**Ejercicio 1: Capacidad de Silos de Granos**

**Planteamiento:**
Se requiere calcular cuántas toneladas de maíz puede almacenar un silo cilíndrico.  
El usuario debe ingresar el radio y la altura del silo en metros.

**Requerimientos:**

1. Solicitar el radio del silo en metros (tipo float) → (r)  
2. Solicitar la altura del silo en metros (tipo float) → (h)  
3. Calcular el volumen del silo:  
   v = π * r² * h  
4. Calcular las toneladas de maíz:  
   t = v * 0.75  

**Prueba del programa:**

Si se ingresan 3 metros de radio y 10 metros de altura, el programa debe informar que la capacidad del silo es de aproximadamente 212.06 toneladas de maíz.

**Analisis E-P-S**

**Entradas:**  
r (radio) [float]  
h (altura) [float]  

**Proceso:**  
v = 3.1416 * r^2 * h  
t = v * 0.75  

**Salida:**  
t (toneladas de maíz) [float]



```python
# ENTRADA: Solicitar radio (r) y altura (h) del silo (tipo float)
r = float(input("Ingrese el radio del silo en metros: "))
h = float(input("Ingrese la altura del silo en metros: "))

# PROCESO: calcular volumen (v) y toneladas (t)
v = 3.1416 * (r**2) * h
t = v * 0.75

# SALIDA: mostrar la cantidad de toneladas almacenadas
print("La capacidad del silo es:", round(t,2), "toneladas de maíz")
```

**Documentacion tecnica**

Línea 1 y 2: Se usa input() para capturar los datos ingresados por el usuario (radio y altura) y float() para convertirlos en números decimales, permitiendo realizar operaciones matemáticas.

Línea 3: Se utiliza el operador ** para elevar el radio al cuadrado y el operador * para multiplicar por 3.1416 y la altura, calculando así el volumen del silo.

Línea 4: Se usa el operador * para multiplicar el volumen por 0.75 y obtener la cantidad de toneladas de maíz.

Línea 5: La función print() muestra el resultado final en la consola, indicando la capacidad del silo en toneladas de maíz, y la función round() se utiliza para redondear el resultado a dos decimales.

**Prueba de escritorio**

Si se ingresan los siguientes datos:

Radio: 3 metros

Altura: 10 metros

Volumen del silo:

v = 3.1416 × (3)² × 10 = 282.74 m³
Toneladas de maíz:

t = 282.74 × 0.75 = 212.06 toneladas
