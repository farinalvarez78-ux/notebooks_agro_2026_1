#**Ejercicio 3: Rendimiento de Pulpa en Frutas**

**Planteamiento:**

Una planta procesadora necesita saber el porcentaje de aprovechamiento de la fruta.
El usuario debe ingresar el peso total de la fruta comprada y el peso de los residuos (cáscaras y semillas).

**Requerimientos:**

1. Solicitar el peso total de la fruta (tipo float) → (pt)

2. Solicitar el peso de los residuos (tipo float) → (r)

3. Calcular la cantidad de pulpa:
p = pt - r

4. Calcular el porcentaje de rendimiento:
pr = (p / pt) * 100

**Prueba del programa:**

Si se ingresan 500 kg de fruta total y 150 kg de residuos, el programa debe informar que se obtienen 350 kg de pulpa con un rendimiento del 70%.

**Analisis E-P-S**

**Entradas:**

pt (peso total) [float]

r (residuos) [float]

**Proceso:**

p = pt - r

pr = (p / pt) * 100

**Salida:**

p (pulpa) [float]

pr (porcentaje de rendimiento) [float]

```python
# ENTRADA: Solicitar peso total (pt) y residuos (r) (tipo float)
pt = float(input("Ingrese el peso total de la fruta en kg: "))
r = float(input("Ingrese el peso de los residuos en kg: "))

# PROCESO: calcular pulpa (p) y rendimiento (pr)
p = pt - r
pr = (p / pt) * 100

# SALIDA: mostrar la cantidad de pulpa y el porcentaje de rendimiento
print("Se obtienen:", int(p), "kg de pulpa")
print("El rendimiento es:", int(pr), "%")
```

**Documentacion tecnica**

Línea 1 y 2: Se usa input() para capturar los datos ingresados por el usuario (peso total y residuos) y float() para convertirlos en números decimales, permitiendo realizar operaciones matemáticas.

Línea 3: Se utiliza el operador - para restar los residuos al peso total y obtener la cantidad de pulpa.

Línea 4: Se utilizan los operadores / y * para calcular el porcentaje de rendimiento, dividiendo la pulpa entre el peso total y multiplicando por 100.

Línea 5 y 6: La función print() muestra los resultados en la consola; se usa int() para convertir los valores a números enteros, eliminando los decimales tanto en la cantidad de pulpa como en el porcentaje de rendimiento.

**Prueba de escritorio**

Si se ingresan los siguientes datos:

Peso total: 500 kg

Residuos: 150 kg

Cantidad de pulpa:

p = 500 - 150 = 350 kg

Porcentaje de rendimiento:

pr = (350 / 500) × 100 = 70 %
