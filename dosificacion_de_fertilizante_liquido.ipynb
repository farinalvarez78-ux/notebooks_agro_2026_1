#**Ejercicio 2: Dosificación de Fertilizante Líquido**

**Planteamiento:**

Un sistema de riego por goteo requiere aplicar una dosis de fertilizante por hectárea.
El usuario debe ingresar las hectáreas totales y la dosis requerida en litros por hectárea.

**Requerimientos:**

1. Solicitar las hectáreas (tipo float) → (h)

2. Solicitar la dosis en litros por hectárea (tipo float) → (d)

3. Calcular la cantidad de fertilizante:f = h * d

4. Calcular la cantidad de agua (relación 1:500):a = f * 500

**Prueba del programa:**

Si se ingresan 5 hectáreas y una dosis de 2 L/ha, el programa debe informar que se necesitan 10 litros de fertilizante y 5000 litros de agua.

**Analisis E-P-S**

**Entradas:**

h (hectáreas) [float]
d (dosis L/ha) [float]

**Proceso:**

f = h * d
a = f * 500

**Salida:**

f (fertilizante) [float]
a (agua) [float]



```python
# ENTRADA: Solicitar hectáreas (h) y dosis (d) (tipo float)
h = float(input("Ingrese la cantidad de hectáreas: "))
d = float(input("Ingrese la dosis en L/ha: "))

# PROCESO: calcular fertilizante (f) y agua (a)
f = h * d
a = f * 500

# SALIDA: mostrar la cantidad de fertilizante y agua requerida
print("Se necesitan:", int(f), "L de producto")
print("Se necesitan:", int(a), "L de agua total")
```

**Documentacion tecnica**

Línea 1 y 2: Se usa input() para capturar los datos ingresados por el usuario (hectáreas y dosis) y float() para convertirlos en números decimales, permitiendo realizar operaciones matemáticas.

Línea 3: Se usa el operador * para multiplicar las hectáreas por la dosis y obtener la cantidad de fertilizante.

Línea 4: Se usa el operador * para multiplicar la cantidad de fertilizante por 500 y calcular la cantidad de agua necesaria según la relación dada.

Línea 5 y 6: La función print() muestra los resultados en la consola, y la función int() se utiliza para convertir los valores a números enteros, eliminando los decimales para una presentación más clara de los resultados.

**Prueba de escritorio**

Si se ingresan los siguientes datos:

Hectáreas: 5
Dosis: 2 L/ha

Cantidad de fertilizante:

f = 5 × 2 = 10 litros

Cantidad de agua:

a = 10 × 500 = 5000 litros
