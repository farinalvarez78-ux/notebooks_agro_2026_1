### 1. Cálculo del Índice de Madurez

En la agroindustria, determinar el momento exacto de cosecha es vital. Este código calcula la relación entre sólidos solubles ($^\circ Brix$) y acidez.

```
# Datos de la muestra
grados_brix = 12.5
acidez_titulable = 0.85

# Cálculo del índice de madurez
indice_madurez = grados_brix / acidez_titulable

print(f"El índice de madurez es: {indice_madurez:.2f}")

if indice_madurez > 12:
    print("Estado: Óptimo para procesamiento.")
else:
    print("Estado: Aún le falta maduración.")
```

### 2. Balance de Masa Simple (Mezclado)

Imagina que necesitas estandarizar el contenido de grasa en una mezcla láctea.

```
# Queremos 100 kg de leche al 3% de grasa
# Tenemos leche al 4% (L1) y leche desnatada al 0.1% (L2)

target_grasa = 3.0
peso_total = 100

# Usando una lógica simple de balance
leche_entera = (target_grasa - 0.1) / (4 - 0.1) * peso_total
leche_desnatada = peso_total - leche_entera

print(f"Para 100kg al 3%, mezcla:")
print(f"- {leche_entera:.2f} kg de leche al 4%")
print(f"- {leche_desnatada:.2f} kg de leche al 0.1%")
```

### 3. Monitoreo de Temperatura en Almacenamiento

Simulación de un sensor que alerta si la cadena de frío se rompe en un silo o bodega.

```
temperaturas = [4.5, 4.8, 5.2, 6.1, 7.5, 5.0] # Lecturas del sensor
limite_max = 6.0

for t in temperaturas:
    if t > limite_max:
        print(f"¡ALERTA! Temperatura crítica detectada: {t}°C")
    else:
        print(f"Temperatura estable: {t}°C")
```
